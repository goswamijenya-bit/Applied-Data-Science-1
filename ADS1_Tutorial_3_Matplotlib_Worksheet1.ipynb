#Exercise 1 - First Line Plot

```python
# 🚀 Run this cell first (libraries + sample data helpers)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Inline plotting for Jupyter and Google Colab
%matplotlib inline

# Utility: deterministic randomness
rng = np.random.default_rng(42)

print('Libraries ready! NumPy:', np.__version__, '| pandas:', pd.__version__)
```

```python
# TODO: Make a basic line plot with labels and title
x = [0,1,2,3,4]
y = [0,1,4,9,16]

# Your code here:
plt.plot(x,y)
plt.xlabel("X")
plt.ylabel("Y")
plt.title(" X vs Y")
#plt.show() - no need to use this if "%matplotlib inline" is used
```

# Exercise - 3 Multiple lines & Legend

```python
# TODO: Plot two lines with legend and grid
x = np.arange(0, 6)
y1 = x
y2 = x**2

# Your code here
plt.plot(x, y1, label = "linear", color = "red", linestyle = ":", linewidth = 6)
plt.plot(x, y2, label = "quadratic", color = "blue", linestyle = "-.", linewidth = 2)
plt.title("Linear and Quadratic")
plt.legend()
plt.grid(True)
plt.show()
```

# Exercise - 4 Scatter Plot & Markers

```python
# TODO: Make a scatter plot with random data and noise
n = 50
x = rng.uniform(0, 10, size=n)
noise = rng.normal(0, 2, size=n)
y = 2*x + 1 + noise

# Your code here
plt.scatter(x, y, marker = "o", color = "red", linestyle = "--", linewidth = 1)
plt.xlabel("X"); plt.ylabel("Y")
plt.title("Scatter & Marker")
plt.show()
```

# Exercise - 6 Histogram

```python
# TODO: Histogram with density
samples = rng.normal(0, 1, size=1000)

# Your code here
plt.hist(samples, bins=30, density=True)
plt.axvline(samples.mean(), linestyle='--')
plt.title('Normal(0,1) Histogram')
plt.show()
```

# Exercise - 10 Fill Between (Confidence Band)

```python
# TODO: Plot a line with a confidence band using fill_between
x = np.linspace(0, 10, 200)
noise = rng.normal(0, 0.2, size=x.size)
y = np.sin(x) + noise
sigma = 0.3

# Your code here
plt.plot(x, y, label='signal')
plt.fill_between(x, y - sigma, y + sigma, alpha=0.3, label='±1σ')
plt.legend()
plt.show()
```
